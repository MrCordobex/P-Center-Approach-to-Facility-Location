```python
import pandas as pd
from pathlib import Path
# Ruta a tu CSV (ajústala a donde lo hayas guardado)

ruta_csv = Path("..") / "data" / "raw" / "20251107_Catalogo_de_Hospitales.csv"

hosp = pd.read_csv(
    ruta_csv,
    sep=",",
    encoding="utf-8-sig"
)

display(hosp)
```

```python
hosp_andalucia = hosp[hosp["Comunidad Autónoma"].astype(str).str.strip() == "Andalucía"].copy()

display(hosp_andalucia)
```

```python
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
import numpy as np

# Suponiendo que ya tienes hosp_andalucia filtrado

# 1. Limpiar código postal
hosp_andalucia["Código Postal"] = (
    hosp_andalucia["Código Postal"]
    .astype(str)
    .str.split(';').str[0]
    .str.strip()
)

# 2. Geocodificador
geolocator = Nominatim(user_agent="andalucia_hosp_cp_geocoder", timeout=10)
geocode = RateLimiter(
    geolocator.geocode,
    min_delay_seconds=1,
    max_retries=2,
    error_wait_seconds=2,
    swallow_exceptions=True
)

# 3. Función con la lógica pedida:
#   1) Nombre + CP + Municipio + Provincia + España
#   2) Dirección + CP + Municipio + Provincia + España
#   3) Si nada -> NaN
def get_coords_two_steps(row):
    nombre = str(row["Nombre"]).strip()
    direccion = str(row["Dirección"]).strip()
    mun = str(row["Municipio"]).strip()
    prov = str(row["Provincia"]).strip()
    cp = str(row["Código Postal"]).strip()

    # 1) Nombre + CP + Municipio + Provincia + España
    if nombre and cp:
        q1 = f"{nombre}, {cp}, {mun}, {prov}, España"
        loc = geocode(q1)
        if loc:
            print(f"Geocoded (1): {q1} -> ({loc.latitude}, {loc.longitude})")
            return pd.Series({"latitud": loc.latitude, "longitud": loc.longitude})

    # 2) Dirección + CP + Municipio + Provincia + España
    if direccion and cp:
        q2 = f"{direccion}, {cp}, {mun}, {prov}, España"
        loc = geocode(q2)
        if loc:
            print(f"Geocoded (2): {q2} -> ({loc.latitude}, {loc.longitude})")
            return pd.Series({"latitud": loc.latitude, "longitud": loc.longitude})
    
    if direccion:
        q3 = f"{direccion}, {mun}, {prov}, España"
        loc = geocode(q3)
        if loc:
            print(f"Geocoded (2 alt): {q3} -> ({loc.latitude}, {loc.longitude})")
            return pd.Series({"latitud": loc.latitude, "longitud": loc.longitude})
    if cp:
        q4 = f"{cp}, {mun}, {prov}, España"
        loc = geocode(q4)
        if loc:
            print(f"Geocoded (2 alt cp): {q4} -> ({loc.latitude}, {loc.longitude})")
            return pd.Series({"latitud": loc.latitude, "longitud": loc.longitude})
    # 3) Si no encuentra en ninguno de los dos casos
    print(f"Could not geocode: Nombre='{nombre}', Dirección='{direccion}', CP='{cp}', Municipio='{mun}', Provincia='{prov}'")
    return pd.Series({"latitud": np.nan, "longitud": np.nan})

# 4. Aplicar función
coords = hosp_andalucia.apply(get_coords_two_steps, axis=1)

# 5. Unir resultados
hosp_andalucia_geo = pd.concat(
    [hosp_andalucia.reset_index(drop=True), coords.reset_index(drop=True)],
    axis=1
)

# 6. Mostrar resultado y contar NaN
display(hosp_andalucia_geo[["Nombre", "Municipio", "Provincia", "Código Postal", "latitud", "longitud"]])

num_nan = hosp_andalucia_geo["latitud"].isna().sum()
print("Filas sin coordenadas (NaN):", num_nan)
```

```python
# Partimos de hosp_andalucia_geo ya creado

# 1. Elegir columna de localidad (según cómo se llame en tu CSV)
loc_col = "Localidad" if "Localidad" in hosp_andalucia_geo.columns else "Municipio"

# 2. Asegurar que las camas son numéricas
hosp_andalucia_geo["Número de camas instaladas"] = (
    pd.to_numeric(hosp_andalucia_geo["Número de camas instaladas"], errors="coerce")
    .fillna(0)
)

# 3. Construir el DataFrame final SIN agrupar hospitales
hospitales_andalucia_final = pd.DataFrame({
    "nombre": hosp_andalucia_geo["Nombre"],
    "localidad": hosp_andalucia_geo[loc_col],
    "provincia": hosp_andalucia_geo["Provincia"],
    "capacidad": hosp_andalucia_geo["Número de camas instaladas"],
    "latitud": hosp_andalucia_geo["latitud"],
    "longitud": hosp_andalucia_geo["longitud"]
})

display(hospitales_andalucia_final)
print("Total filas:", hospitales_andalucia_final.shape[0])
```

```python
ruta_destino_hosp = Path("..") / "data" / "processed" / "hospitales_coordenadas_andalucia.csv"
hospitales_andalucia_final.to_csv(ruta_destino_hosp, index=False)
```
